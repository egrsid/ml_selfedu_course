```python
import numpy as np

TP = [45, 37, 51, 47]
TN = [36, 37, 29, 28]
FP = [18, 21, 15, 17]
FN = [8, 11, 9, 5]

tpm = np.mean(TP)
tnm = np.mean(TN)
fpm = np.mean(FP)
fnm = np.mean(FN)

precision = tpm / (tpm + fpm)
recall = tpm / (tpm + fnm)
```
